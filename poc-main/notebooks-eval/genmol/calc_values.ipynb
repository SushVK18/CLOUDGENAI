```python
from rdkit import Chem
from rdkit.Chem import Descriptors
import numpy as np
import pandas as pd
```

```python
def calculate_logp_and_mw(smiles):
    # Convert SMILES to molecule
    mol = Chem.MolFromSmiles(smiles)
    
    if mol is None:
        raise ValueError("Invalid SMILES string.")
    
    # Calculate logP and molecular weight
    logp = Descriptors.MolLogP(mol)
    mw = Descriptors.MolWt(mol)
    
    return logp, mw

def calculate_statistics(values):
    mean = np.mean(values)
    std_dev = np.std(values)
    second_std_dev = std_dev * 2
    return mean, std_dev, second_std_dev


```

```python
df_bayer = pd.read_csv(r"../../data/raw/df_ICA_bionemo_eval_20250219.csv", sep='\t')
```

```python
df_bayer.sort_values(by="AI_H", ascending=False, inplace=True)
df_bayer.head(20)
```

```python
df_bayer[:100].shape
```

```python
df_top = df_bayer[:100].copy()
smiles_list_bayer = df_top["std_smiles"]
```

```python
logp_values = []
mw_values = []

for smiles in smiles_list_bayer:
    logp, mw = calculate_logp_and_mw(smiles)
    logp_values.append(logp)
    mw_values.append(mw)

# Calculate statistics for logP and MW
logp_stats = calculate_statistics(logp_values)
mw_stats = calculate_statistics(mw_values)

#print(f"LogP values: {logp_values}")
print(f"LogP Mean: {logp_stats[0]}, Std Dev: {logp_stats[1]}, 2nd Std Dev: {logp_stats[2]}")
#print(f"MW values: {mw_values}")
print(f"MW Mean: {mw_stats[0]}, Std Dev: {mw_stats[1]}, 2nd Std Dev: {mw_stats[2]}")
```

```python
print(f"LogP Mean + STD: {logp_stats[0] + logp_stats[1]}, LogP Mean - STD: { logp_stats[0] - logp_stats[1]}")
print(f"LogP Mean + 2nd STD: {logp_stats[0] + logp_stats[2]}, LogP Mean - 2nd STD: { logp_stats[0] - logp_stats[2]}")

print(f"LogP Mean + STD: {mw_stats[0] + mw_stats[1]}, LogP Mean - STD: { mw_stats[0] - mw_stats[1]}")
print(f"LogP Mean + 2nd STD: {mw_stats[0] + mw_stats[2]}, LogP Mean - 2nd STD: { mw_stats[0] - mw_stats[2]}")
```
