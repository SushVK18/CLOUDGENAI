```python
# Import the module
from alphafold_client import AlphaFoldBatchProcessor


# Create a processor instance
processor = AlphaFoldBatchProcessor(
    af2_host='http://localhost:8081',
)

# Process a directory of FASTA files
results = processor.process_directory(
    input_dir='../../data/external/uniprot/20250316_110329_uniprot_entries_16',
    output_dir='../../data/interim/af2'
)

# Access results as before
print(f"Processed {results['total']} files: {results['success']} successful, {results['failure']} failed")
```
