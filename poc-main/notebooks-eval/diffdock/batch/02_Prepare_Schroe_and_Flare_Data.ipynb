## Cross Eval - with a quick visualization

```python
from rdkit import Chem
from rdkit.Chem import AllChem
import py3Dmol

test_sdf = "../../../data/interim/docking/ligands_from_prep_Flare_SDF/bcsin_226_ligands.sdf"



# Load molecule from 3D SDF file
mol = Chem.SDMolSupplier(test_sdf, removeHs=False)[0]  # Gets first molecule, keeps hydrogens

# Create a 3D visualization with py3Dmol
view = py3Dmol.view(width=400, height=400)
mb = Chem.MolToMolBlock(mol)
view.addModel(mb, 'sdf')
view.setStyle({'stick':{}})
view.zoomTo()
view.show()
```
