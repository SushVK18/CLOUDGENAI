```python
#### results Flare+Flare
# Directories
prep_pdb_dir = r"../../../data/interim/docking/prep_Schroe"
#sdf_base_dir = r"../../../data/interim/docking/Res_Diffdock_Schroe_R1" # old version with sdf files diffdock used
ligand_base_dir = r"../../../data/raw/docking/Results_Flare_Flare/" # new path with results from manual docking
gt_pdb_dir = r"../../../data/raw/docking/GT" 

combined_results = []
### RUN ANALYSIS
# Calculate RMSD for all pairs
combined_results = batch_calculate_rmsd(
    pairs_df,
    prep_pdb_dir,
    ligand_base_dir,
    gt_pdb_dir,
    file_pattern=None,
    save_superimposed=False
)
```
