```python
import py3Dmol
import re
from IPython.display import display

def visualize_protein_ligand(pdb_file, ligand_id=None, width=800, height=500):
    """
    Visualize a protein-ligand complex with py3Dmol
    
    Parameters:
    -----------
    pdb_file : str
        Path to the PDB file containing the protein-ligand complex
    ligand_id : str, optional
        Specific ligand ID to use. If None, will try to detect the organic ligand
    width : int
        Width of the viewer
    height : int
        Height of the viewer
    """
    # Create a py3Dmol viewer
    view = py3Dmol.view(width=width, height=height)
    
    # Load the PDB file
    with open(pdb_file, 'r') as f:
        pdb_data = f.read()
    
    view.addModel(pdb_data, 'pdb')
    
    # Define metal ions and common cofactors to exclude
    metal_ions = ["MN", "ZN", "CD", "FE", "MG", "CA", "NA", "K", "CU", "NI", "CO"]
    common_cofactors = ["HOH", "WAT", "SO4", "PO4", "GOL", "CL"]
    
    # If ligand_id is provided, use it directly
    if ligand_id:
        # Set the protein as cartoon
        view.setStyle({'and': [{'not': {'resn': ligand_id}}, 
                               {'not': {'resn': common_cofactors}}]}, 
                      {'cartoon': {'color': 'spectrum'}})
        
        # Set the ligand as sticks
        view.setStyle({'resn': ligand_id}, 
                      {'stick': {'colorscheme': 'greenCarbon', 'radius': 0.2}})
    else:
        # No ligand ID specified, try to find the ligand
        
        # First set all HETATMs as sticks
        view.setStyle({'hetflag': True, 'not': {'resn': common_cofactors + metal_ions}}, 
                      {'stick': {'colorscheme': 'greenCarbon', 'radius': 0.2}})
        
        # Set protein as cartoon
        view.setStyle({'hetflag': False}, {'cartoon': {'color': 'spectrum'}})
        
        # Try to identify ligands with 3-digit numeric IDs (typical for docked ligands)
        # This is similar to the logic in your original code
        for line in pdb_data.split('\n'):
            if line.startswith('HETATM'):
                res_name = line[17:20].strip()
                # Check if it's a three-digit number (typical for docked ligands)
                if re.match(r'^\d{3}$', res_name) and res_name not in common_cofactors:
                    # Highlight this ligand differently
                    view.setStyle({'resn': res_name}, 
                                 {'cartoon': {'color': 'white'}}) 
                                  #{'stick': {'colorscheme': 'magentaCarbon', 'radius': 0.3}})
                    #print(f"Highlighted potential ligand with ID: {res_name}")
    
    # Show metal ions as spheres
    view.setStyle({'resn': metal_ions}, {'sphere': {'colorscheme': 'grayCarbon', 'radius': 0.5}})
    
    # Add surface representation with transparency
    #view.addSurface(py3Dmol.VDW, {'opacity': 0.7, 'color': 'white'})
    
    # Set background color and other viewing options
    view.setBackgroundColor('white')
    view.zoomTo()
    view.zoom(0.8)
    
    return view

# Example usage
# view = visualize_protein_ligand("complex_result2.pdb")
# view.show()
```

```python
def extract_ligand_smiles(pdb_file, ligand_id=None, chain_id=None, res_num=None):
    """
    Extract SMILES representation of a ligand from a PDB file
    
    Parameters:
    -----------
    pdb_file : str
        Path to the PDB file containing the protein-ligand complex
    ligand_id : str, optional
        Specific ligand ID to use. If None, will try to detect the organic ligand
    chain_id : str, optional
        Chain identifier to select a specific instance of the ligand
    res_num : str, optional
        Residue number to select a specific instance of the ligand
        
    Returns:
    --------
    str
        SMILES representation of the ligand
    dict
        Additional information about the extraction
    """
    # Read PDB file
    with open(pdb_file, 'r') as f:
        pdb_lines = f.readlines()
    
    # Define metal ions and common cofactors to exclude
    metal_ions = ["MN", "ZN", "CD", "FE", "MG", "CA", "NA", "K", "CU", "NI", "CO"]
    common_cofactors = ["HOH", "WAT", "SO4", "PO4", "GOL", "CL"]
    
    # Group HETATM records by residue name, chain, and number
    ligand_groups = {}
    
    for line in pdb_lines:
        if line.startswith('HETATM'):
            res_name = line[17:20].strip()
            current_chain_id = line[21]
            current_res_num = line[22:26].strip()
            
            # Skip water, ions, and known cofactors
            if res_name in common_cofactors or res_name in metal_ions:
                continue
            
            # Create a unique key for this residue
            key = (res_name, current_chain_id, current_res_num)
            
            if key not in ligand_groups:
                ligand_groups[key] = []
            
            ligand_groups[key].append(line)
    
    # If specific identifiers are provided, use them
    if ligand_id and chain_id and res_num:
        selected_key = (ligand_id, chain_id, res_num)
        if selected_key in ligand_groups:
            selected_ligand = selected_key
        else:
            available = [f"{k[0]} (chain {k[1]}, res {k[2]}, atoms: {len(v)})" 
                         for k, v in ligand_groups.items()]
            raise ValueError(f"Specified ligand {selected_key} not found. Available: {available}")
    
    # If just ligand_id is provided, find all instances
    elif ligand_id:
        matching_keys = [key for key in ligand_groups.keys() if key[0] == ligand_id]
        if not matching_keys:
            available_residues = list(set(key[0] for key in ligand_groups.keys()))
            raise ValueError(f"No ligand with ID {ligand_id} found. Available: {available_residues}")
        
        # Use the instance with the most atoms
        selected_ligand = max(matching_keys, key=lambda k: len(ligand_groups[k]))
        
        # Warn if multiple instances found
        if len(matching_keys) > 1:
            instances = [f"Chain {k[1]}, Res {k[2]} ({len(ligand_groups[k])} atoms)" 
                        for k in matching_keys]
            print(f"Warning: Multiple instances of ligand {ligand_id} found: {instances}")
            print(f"Selected: Chain {selected_ligand[1]}, Res {selected_ligand[2]} "
                  f"({len(ligand_groups[selected_ligand])} atoms)")
    
    # No specific identifiers provided, find the largest organic molecule
    else:
        organic_keys = []
        for key, lines in ligand_groups.items():
            # Check if this residue has carbon atoms (organic)
            has_carbon = any('C' in line[76:78].strip() for line in lines if len(line) >= 78)
            if has_carbon:
                organic_keys.append(key)
        
        if not organic_keys:
            available = [f"{k[0]} (chain {k[1]}, res {k[2]})" for k in ligand_groups.keys()]
            raise ValueError(f"No organic ligands found. Available hetero groups: {available}")
        
        # Use the largest organic molecule
        selected_ligand = max(organic_keys, key=lambda k: len(ligand_groups[k]))
        print(f"Auto-selected ligand: {selected_ligand[0]} (Chain {selected_ligand[1]}, "
              f"Res {selected_ligand[2]}, {len(ligand_groups[selected_ligand])} atoms)")
    
    # Extract the selected ligand
    ligand_atoms = ligand_groups[selected_ligand]
    
    if not ligand_atoms:
        raise ValueError(f"No atoms found for selected ligand")
    
    # Create a temporary PDB file with only the ligand
    import tempfile
    with tempfile.NamedTemporaryFile(suffix='.pdb', delete=False) as tmp:
        tmp_name = tmp.name
        for line in ligand_atoms:
            tmp.write(line.encode())
    
    # Use RDKit to read the ligand PDB and convert to SMILES
    try:
        # Try to read the molecule with RDKit
        mol = Chem.MolFromPDBFile(tmp_name, removeHs=False)
        
        if mol is None:
            # If standard reading fails, try using more options
            mol = Chem.MolFromPDBFile(tmp_name, 
                                      removeHs=False,
                                      sanitize=False)
            
            if mol is not None:
                # Try to sanitize the molecule
                try:
                    Chem.SanitizeMol(mol)
                except:
                    print("Warning: Molecule sanitization failed, SMILES may be incorrect")

        if mol is None:
            raise ValueError(f"RDKit couldn't parse the selected ligand")
        
        # Generate SMILES
        smiles = Chem.MolToSmiles(mol)
        
        # Clean up the temporary file
        import os
        os.unlink(tmp_name)
        
        # Return information
        info = {
            "ligand_id": selected_ligand[0],
            "chain_id": selected_ligand[1],
            "res_num": selected_ligand[2],
            "atom_count": len(ligand_atoms),
            "mol_object": mol  # Return the RDKit mol object for further use if needed
        }
        
        return smiles, info
        
    except Exception as e:
        # Clean up the temporary file
        import os
        os.unlink(tmp_name)
        raise ValueError(f"Error extracting SMILES: {str(e)}")
```

```python
def extract_ligand_smiles(pdb_file, ligand_id=None, chain_id=None, res_num=None, altloc=None):
    """
    Extract SMILES representation of a ligand from a PDB file
    
    Parameters:
    -----------
    pdb_file : str
        Path to the PDB file containing the protein-ligand complex
    ligand_id : str, optional
        Specific ligand ID to use. If None, will try to detect the organic ligand
    chain_id : str, optional
        Chain identifier to select a specific instance of the ligand
    res_num : str, optional
        Residue number to select a specific instance of the ligand
    altloc : str, optional
        Alternate location identifier (A, B, etc.) if multiple conformations exist
        
    Returns:
    --------
    str
        SMILES representation of the ligand
    dict
        Additional information about the extraction
    """
    # Read PDB file
    with open(pdb_file, 'r') as f:
        pdb_lines = f.readlines()
    
    # Define metal ions and common cofactors to exclude
    metal_ions = ["MN", "ZN", "CD", "FE", "MG", "CA", "NA", "K", "CU", "NI", "CO"]
    common_cofactors = ["HOH", "WAT", "SO4", "PO4", "GOL", "CL"]
    
    # Group HETATM records by residue name, chain, number, and altloc
    ligand_groups = {}
    
    for line in pdb_lines:
        if line.startswith('HETATM'):
            res_name = line[17:20].strip()
            current_chain_id = line[21]
            current_res_num = line[22:26].strip()
            current_altloc = line[16].strip()  # Extract alternate location indicator
            
            # Skip water, ions, and known cofactors
            if res_name in common_cofactors or res_name in metal_ions:
                continue
            
            # Extract occupancy value
            try:
                occupancy = float(line[54:60].strip())
            except ValueError:
                occupancy = 1.0
            
            # Create a unique key for this residue
            key = (res_name, current_chain_id, current_res_num, current_altloc)
            
            if key not in ligand_groups:
                ligand_groups[key] = {"lines": [], "occupancy": occupancy}
            
            ligand_groups[key]["lines"].append(line)
    
    # Check if we found any ligands
    if not ligand_groups:
        raise ValueError(f"No ligands found in {pdb_file}")
    
    # Print debugging information about found ligands
    print("Potential ligands found:")
    print("ID  | Chain | Res# | Alt | Atoms | Occupancy")
    print("----|-------|------|-----|-------|----------")
    for (lid, cid, rnum, alt), data in ligand_groups.items():
        alt_display = alt if alt else " "
        print(f"{lid:<4} | {cid:<5} | {rnum:<4} | {alt_display:<3} | {len(data['lines']):<5} | {data['occupancy']:.2f}")
    
    # If specific identifiers are provided, use them
    if ligand_id and chain_id and res_num:
        matching_keys = []
        for key in ligand_groups.keys():
            if key[0] == ligand_id and key[1] == chain_id and key[2] == res_num:
                if altloc is None or key[3] == altloc:
                    matching_keys.append(key)
        
        if not matching_keys:
            raise ValueError(f"Specified ligand {ligand_id} {chain_id} {res_num} not found.")
        
        # If multiple alternates and no altloc specified, use the one with highest occupancy
        if len(matching_keys) > 1 and altloc is None:
            selected_key = max(matching_keys, 
                               key=lambda k: ligand_groups[k]["occupancy"])
            print(f"Multiple alternates found, using highest occupancy: {selected_key[3]}")
        else:
            selected_key = matching_keys[0]
    
    # If just ligand_id is provided, find all instances
    elif ligand_id:
        matching_keys = [key for key in ligand_groups.keys() if key[0] == ligand_id]
        if not matching_keys:
            available_residues = list(set(key[0] for key in ligand_groups.keys()))
            raise ValueError(f"No ligand with ID {ligand_id} found. Available: {available_residues}")
        
        # Group by chain and residue number (ignoring altloc)
        grouped_keys = {}
        for key in matching_keys:
            group_key = (key[0], key[1], key[2])  # ligand_id, chain_id, res_num
            if group_key not in grouped_keys:
                grouped_keys[group_key] = []
            grouped_keys[group_key].append(key)
        
        # For each group, select the alternate with highest occupancy
        best_keys = []
        for group, keys in grouped_keys.items():
            best_key = max(keys, key=lambda k: ligand_groups[k]["occupancy"])
            best_keys.append(best_key)
        
        # Use the instance with the most atoms
        selected_key = max(best_keys, 
                          key=lambda k: len(ligand_groups[k]["lines"]))
        
        # Warn if multiple instances found
        if len(best_keys) > 1:
            instances = [f"Chain {k[1]}, Res {k[2]} ({len(ligand_groups[k]['lines'])} atoms)" 
                        for k in best_keys]
            print(f"Multiple instances of ligand {ligand_id} found: {instances}")
            print(f"Selected: Chain {selected_key[1]}, Res {selected_key[2]}, "
                  f"Alt {selected_key[3] if selected_key[3] else 'none'} "
                  f"({len(ligand_groups[selected_key]['lines'])} atoms)")
    
    # No specific identifiers provided, find the largest organic molecule
    else:
        # Filter for organic molecules (containing carbon)
        organic_keys = []
        for key, data in ligand_groups.items():
            # Check if this residue has carbon atoms
            has_carbon = any('C' in line[76:78].strip() 
                            for line in data["lines"] if len(line) >= 78)
            if has_carbon:
                organic_keys.append(key)
        
        if not organic_keys:
            raise ValueError(f"No organic ligands found.")
        
        # Group by residue name, chain, and number (ignoring altloc)
        grouped_keys = {}
        for key in organic_keys:
            group_key = (key[0], key[1], key[2])  # ligand_id, chain_id, res_num
            if group_key not in grouped_keys:
                grouped_keys[group_key] = []
            grouped_keys[group_key].append(key)
        
        # For each group, select the alternate with highest occupancy
        best_keys = []
        for group, keys in grouped_keys.items():
            best_key = max(keys, key=lambda k: ligand_groups[k]["occupancy"])
            best_keys.append(best_key)
        
        # Use the largest organic molecule
        selected_key = max(best_keys, 
                          key=lambda k: len(ligand_groups[k]["lines"]))
        
        print(f"Auto-selected ligand: {selected_key[0]} (Chain {selected_key[1]}, "
              f"Res {selected_key[2]}, Alt {selected_key[3] if selected_key[3] else 'none'}, "
              f"{len(ligand_groups[selected_key]['lines'])} atoms)")
    
    # Extract the selected ligand
    ligand_atoms = ligand_groups[selected_key]["lines"]
    
    if not ligand_atoms:
        raise ValueError(f"No atoms found for selected ligand")
    
    # Create a temporary PDB file with only the ligand
    import tempfile
    with tempfile.NamedTemporaryFile(suffix='.pdb', delete=False) as tmp:
        tmp_name = tmp.name
        for line in ligand_atoms:
            tmp.write(line.encode())
    
    # Use RDKit to read the ligand PDB and convert to SMILES
    try:
        # Try to read the molecule with RDKit
        mol = Chem.MolFromPDBFile(tmp_name, removeHs=False)
        
        if mol is None:
            # If standard reading fails, try using more options
            mol = Chem.MolFromPDBFile(tmp_name, 
                                      removeHs=False,
                                      sanitize=False)
            
            if mol is not None:
                # Try to sanitize the molecule
                try:
                    Chem.SanitizeMol(mol)
                except:
                    print("Warning: Molecule sanitization failed, SMILES may be incorrect")

        if mol is None:
            raise ValueError(f"RDKit couldn't parse the selected ligand")
        
        # Generate SMILES
        smiles = Chem.MolToSmiles(mol)
        
        # Clean up the temporary file
        import os
        os.unlink(tmp_name)
        
        # Return information
        info = {
            "ligand_id": selected_key[0],
            "chain_id": selected_key[1],
            "res_num": selected_key[2],
            "atom_count": len(ligand_atoms),
            "mol_object": mol  # Return the RDKit mol object for further use if needed
        }
        
        return smiles, info
        
    except Exception as e:
        # Clean up the temporary file
        import os
        os.unlink(tmp_name)
        raise ValueError(f"Error extracting SMILES: {str(e)}")
```

```python
def extract_ligand_smiles_with_conect(pdb_file, ligand_id=None, chain_id=None, res_num=None):
    """
    Extract SMILES representation of a ligand from a PDB file using CONECT records
    
    Parameters:
    -----------
    pdb_file : str
        Path to the PDB file containing the protein-ligand complex
    ligand_id : str, optional
        Specific ligand ID to use. If None, will try to detect the organic ligand
    chain_id : str, optional
        Chain identifier to select a specific instance of the ligand
    res_num : str, optional
        Residue number to select a specific instance of the ligand
        
    Returns:
    --------
    str
        SMILES representation of the ligand
    dict
        Additional information about the extraction
    """
    from rdkit import Chem
    from rdkit.Chem import AllChem
    import tempfile
    import os
    import numpy as np
    
    # Read PDB file
    with open(pdb_file, 'r') as f:
        pdb_lines = f.readlines()
    
    # Parse HETATM records
    hetero_atoms = {}
    atom_indices = {}  # Maps PDB atom index to atom object
    
    # Define metal ions and common cofactors to exclude
    metal_ions = ["MN", "ZN", "CD", "FE", "MG", "CA", "NA", "K", "CU", "NI", "CO"]
    common_cofactors = ["HOH", "WAT", "SO4", "PO4", "GOL", "CL"]
    
    # First pass: collect all HETATM records
    for line in pdb_lines:
        if line.startswith('HETATM'):
            atom_index = int(line[6:11].strip())
            res_name = line[17:20].strip()
            current_chain_id = line[21]
            current_res_num = line[22:26].strip()
            
            # Skip water, ions, and known cofactors
            if res_name in common_cofactors or res_name in metal_ions:
                continue
            
            # Check if we're looking for a specific ligand
            if ligand_id and res_name != ligand_id:
                continue
            if chain_id and current_chain_id != chain_id:
                continue
            if res_num and current_res_num != res_num:
                continue
            
            key = (res_name, current_chain_id, current_res_num)
            
            if key not in hetero_atoms:
                hetero_atoms[key] = []
            
            hetero_atoms[key].append((atom_index, line))
            atom_indices[atom_index] = (key, len(hetero_atoms[key]) - 1)  # Store reference to the atom
    
    # Second pass: parse CONECT records
    connections = {}
    connected_groups = {}  # Groups of residues connected by CONECT records
    
    for line in pdb_lines:
        if line.startswith('CONECT'):
            parts = line.split()
            if len(parts) < 2:
                continue
                
            try:
                from_atom = int(parts[1])
                to_atoms = [int(p) for p in parts[2:]]
                
                # Store connections
                if from_atom not in connections:
                    connections[from_atom] = []
                connections[from_atom].extend(to_atoms)
                
                # Check if these atoms belong to different residues
                if from_atom in atom_indices:
                    from_res = atom_indices[from_atom][0]  # (res_name, chain_id, res_num)
                    
                    for to_atom in to_atoms:
                        if to_atom in atom_indices:
                            to_res = atom_indices[to_atom][0]
                            
                            if from_res != to_res:
                                # These residues are connected
                                if from_res not in connected_groups:
                                    connected_groups[from_res] = set([from_res])
                                if to_res not in connected_groups:
                                    connected_groups[to_res] = set([to_res])
                                
                                # Merge the groups
                                merged_group = connected_groups[from_res].union(connected_groups[to_res])
                                for res in merged_group:
                                    connected_groups[res] = merged_group
            except ValueError:
                # Skip invalid CONECT records
                continue
    
    # Find groups of connected residues
    unique_groups = []
    processed = set()
    
    for key in hetero_atoms.keys():
        if key in processed:
            continue
            
        if key in connected_groups:
            group = connected_groups[key]
            
            # Check if we already processed this group
            if not any(key in g for g in unique_groups):
                unique_groups.append(group)
                processed.update(group)
        else:
            # Residue not connected to others
            unique_groups.append(set([key]))
            processed.add(key)
    
    # Print information about found residues and their connections
    print(f"Found {len(hetero_atoms)} HETATM residue(s)")
    for key, atoms in hetero_atoms.items():
        res_name, chain, res_num = key
        print(f"  {res_name} (Chain {chain}, Res {res_num}): {len(atoms)} atoms")
    
    print(f"\nFound {len(unique_groups)} connected group(s)")
    for i, group in enumerate(unique_groups):
        print(f"Group {i+1}:")
        atom_count = 0
        for res in group:
            res_name, chain, res_num = res
            atoms = hetero_atoms[res]
            atom_count += len(atoms)
            print(f"  {res_name} (Chain {chain}, Res {res_num}): {len(atoms)} atoms")
        print(f"  Total: {atom_count} atoms")
    
    # Select the group to process
    selected_group = None
    
    # If specific identifiers were provided, find matching group
    if ligand_id or chain_id or res_num:
        for group in unique_groups:
            # Check if any residue in the group matches criteria
            matching = False
            for res in group:
                res_name, res_chain, res_num_val = res
                if ligand_id and res_name != ligand_id:
                    continue
                if chain_id and res_chain != chain_id:
                    continue
                if res_num and res_num_val != res_num:
                    continue
                matching = True
                break
                
            if matching:
                selected_group = group
                break
    
    # If no specific criteria or no match, use the largest group
    if selected_group is None:
        if not unique_groups:
            raise ValueError("No suitable ligands found")
            
        # Find the largest group that contains carbon atoms
        max_size = 0
        for group in unique_groups:
            # Count atoms in this group
            atoms = []
            for res in group:
                atoms.extend(hetero_atoms[res])
                
            # Check if group has carbon atoms
            has_carbon = any('C' in line[76:78].strip() for _, line in atoms if len(line) >= 78)
            
            if has_carbon and len(atoms) > max_size:
                max_size = len(atoms)
                selected_group = group
    
    if selected_group is None:
        raise ValueError("Could not find a suitable ligand")
    
    # Create a temporary PDB file with only the selected ligand group
    with tempfile.NamedTemporaryFile(suffix='.pdb', delete=False) as tmp:
        tmp_name = tmp.name
        
        # Write all atoms from the selected group
        atom_count = 0
        for res in selected_group:
            for idx, line in hetero_atoms[res]:
                tmp.write(line.encode())
                atom_count += 1
        
        # Write CONECT records for atoms in this group
        for res in selected_group:
            for idx, _ in hetero_atoms[res]:
                if idx in connections:
                    # Only include connections to atoms in this group
                    connected_atoms = []
                    for con in connections[idx]:
                        if con in atom_indices and atom_indices[con][0] in selected_group:
                            connected_atoms.append(con)
                    
                    if connected_atoms:
                        conect_line = f"CONECT {idx:5d}"
                        for con in connected_atoms:
                            conect_line += f" {con:5d}"
                        conect_line += "\n"
                        tmp.write(conect_line.encode())
    
    # Use RDKit to read the ligand PDB and convert to SMILES
    try:
        # Try to read the molecule with RDKit
        mol = Chem.MolFromPDBFile(tmp_name, removeHs=False, proximityBonding=False)
        
        if mol is None:
            print("Standard PDB reading failed, trying with proximity bonding...")
            mol = Chem.MolFromPDBFile(tmp_name, 
                                      removeHs=False,
                                      proximityBonding=True,
                                      sanitize=False)
        
        if mol is None:
            print("PDB reading with proximity bonding failed, trying to read only CONECT records...")
            mol = Chem.MolFromPDBFile(tmp_name, 
                                     removeHs=False,
                                     proximityBonding=False,
                                     sanitize=False)
        
        if mol is not None:
            try:
                Chem.SanitizeMol(mol)
            except:
                print("Warning: Molecule sanitization failed, SMILES may be incorrect")

        if mol is None:
            raise ValueError("RDKit couldn't parse the ligand group")
        
        # Generate SMILES
        smiles = Chem.MolToSmiles(mol)
        
        # Clean up the temporary file
        os.unlink(tmp_name)
        
        # Return information
        res_names = [res[0] for res in selected_group]
        chains = [res[1] for res in selected_group]
        res_nums = [res[2] for res in selected_group]
        
        info = {
            "ligand_ids": res_names,
            "chain_ids": chains,
            "res_nums": res_nums,
            "atom_count": atom_count,
            "mol_object": mol  # Return the RDKit mol object for further use if needed
        }
        
        return smiles, info
        
    except Exception as e:
        # Clean up the temporary file
        os.unlink(tmp_name)
        raise ValueError(f"Error extracting SMILES: {str(e)}")
```

```python
def visualize_protein_ligand(pdb_file, ligand_id=None, chain_id=None, res_num=None, width=800, height=500):
    """
    Visualize a protein-ligand complex with py3Dmol
    
    Parameters:
    -----------
    pdb_file : str
        Path to the PDB file containing the protein-ligand complex
    ligand_id : str, optional
        Specific ligand ID to use. If None, will try to detect the organic ligand
    chain_id : str, optional
        Chain identifier to select a specific instance of the ligand
    res_num : str, optional
        Residue number to select a specific instance of the ligand
    width : int
        Width of the viewer
    height : int
        Height of the viewer
    """
    # Create a py3Dmol viewer
    view = py3Dmol.view(width=width, height=height)
    
    # Load the PDB file
    with open(pdb_file, 'r') as f:
        pdb_data = f.read()
    
    view.addModel(pdb_data, 'pdb')
    
    # Define metal ions and common cofactors to exclude
    metal_ions = ["MN", "ZN", "CD", "FE", "MG", "CA", "NA", "K", "CU", "NI", "CO"]
    common_cofactors = ["HOH", "WAT", "SO4", "PO4", "GOL", "CL"]
    
    # If specific identifiers are provided, use them directly
    if ligand_id:
        # Set the protein as cartoon
        view.setStyle({'and': [{'not': {'resn': ligand_id}}, 
                              {'not': {'resn': common_cofactors}}]}, 
                      {'cartoon': {'color': 'spectrum'}})
        
        # Create more specific selector for the ligand
        ligand_selector = {'resn': ligand_id}
        if chain_id:
            ligand_selector['chain'] = chain_id
        if res_num:
            ligand_selector['resi'] = res_num
            
        # Set the ligand as sticks
        view.setStyle(ligand_selector, 
                      {'stick': {'colorscheme': 'greenCarbon', 'radius': 0.2}})
        
        print(f"Highlighting ligand {ligand_id}" + 
              (f" in chain {chain_id}" if chain_id else "") + 
              (f" at position {res_num}" if res_num else ""))
    else:
        # No ligand ID specified, style all HETATMs as sticks except common ones
        view.setStyle({'hetflag': True, 'not': {'resn': common_cofactors + metal_ions}}, 
                      {'stick': {'colorscheme': 'greenCarbon', 'radius': 0.2}})
        
        # Set protein as cartoon
        view.setStyle({'hetflag': False}, {'cartoon': {'color': 'spectrum'}})
    
    # Show metal ions as spheres
    view.setStyle({'resn': metal_ions}, {'sphere': {'colorscheme': 'grayCarbon', 'radius': 0.5}})
    
    # Add surface representation with transparency
    #view.addSurface(py3Dmol.VDW, {'opacity': 0.7, 'color': 'white'})
    
    # Set background color and other viewing options
    view.setBackgroundColor('white')
    view.zoomTo()
    view.zoom(0.8)
    
    return view
```

```python
def split_docking_poses(input_pdb, output_prefix="pose", ligand_id=None, include_waters=True, include_cofactors=True):
    """
    Split a PDB file containing multiple docking poses into separate PDB files.
    
    Parameters:
    -----------
    input_pdb : str
        Path to the input PDB file containing multiple docking poses
    output_prefix : str
        Prefix for output PDB files (default: "pose")
    ligand_id : str, optional
        Specific ligand ID to look for. If None, will try to identify ligands
    include_waters : bool
        Whether to include water molecules in output files (default: True)
    include_cofactors : bool
        Whether to include common cofactors in output files (default: True)
        
    Returns:
    --------
    list
        List of output filenames created
    """
    # Read PDB file
    with open(input_pdb, 'r') as f:
        pdb_lines = f.readlines()
    
    # Define metal ions and common cofactors
    metal_ions = ["MN", "ZN", "CD", "FE", "MG", "CA", "NA", "K", "CU", "NI", "CO"]
    waters = ["HOH", "WAT"]
    cofactors = ["SO4", "PO4", "GOL", "CL"]
    
    # Group HETATM records by models and types
    models = {}
    current_model = 0
    protein_lines = []
    conect_lines = []
    
    # First, identify protein atoms and CONECT records
    for line in pdb_lines:
        if line.startswith('ATOM'):
            protein_lines.append(line)
        elif line.startswith('CONECT'):
            conect_lines.append(line)
    
    # Now process the file to extract models and ligands
    for line in pdb_lines:
        if line.startswith('MODEL'):
            try:
                current_model = int(line.split()[1])
                if current_model not in models:
                    models[current_model] = {
                        'ligands': {},
                        'waters': {},
                        'cofactors': {},
                        'metals': {}
                    }
            except:
                pass
        elif line.startswith('HETATM'):
            res_name = line[17:20].strip()
            chain_id = line[21]
            res_num = line[22:26].strip()
            atom_num = int(line[6:11].strip())
            
            key = (res_name, chain_id, res_num)
            
            if current_model not in models:
                models[current_model] = {
                    'ligands': {},
                    'waters': {},
                    'cofactors': {},
                    'metals': {}
                }
            
            # Categorize the HETATM
            if res_name in waters:
                if include_waters:
                    if key not in models[current_model]['waters']:
                        models[current_model]['waters'][key] = []
                    models[current_model]['waters'][key].append((atom_num, line))
            elif res_name in cofactors:
                if include_cofactors:
                    if key not in models[current_model]['cofactors']:
                        models[current_model]['cofactors'][key] = []
                    models[current_model]['cofactors'][key].append((atom_num, line))
            elif res_name in metal_ions:
                if include_cofactors:  # Include metals with cofactors
                    if key not in models[current_model]['metals']:
                        models[current_model]['metals'][key] = []
                    models[current_model]['metals'][key].append((atom_num, line))
            else:
                # Skip if we're looking for a specific ligand and this isn't it
                if ligand_id and res_name != ligand_id:
                    continue
                
                if key not in models[current_model]['ligands']:
                    models[current_model]['ligands'][key] = []
                models[current_model]['ligands'][key].append((atom_num, line))
    
    # Filter CONECT records for each model
    model_conect_lines = {}
    for model in models:
        model_conect_lines[model] = []
        model_atoms = set()
        
        # Collect all atom numbers in this model
        for category in ['ligands', 'waters', 'cofactors', 'metals']:
            for key, atoms in models[model][category].items():
                for atom_num, _ in atoms:
                    model_atoms.add(atom_num)
        
        # Filter CONECT records for atoms in this model
        for line in conect_lines:
            parts = line.split()
            if len(parts) < 2:
                continue
                
            try:
                from_atom = int(parts[1])
                if from_atom in model_atoms:
                    model_conect_lines[model].append(line)
            except ValueError:
                continue
    
    # Create output files
    output_files = []
    
    for model in sorted(models.keys()):
        if not models[model]['ligands']:
            continue  # Skip models without ligands
            
        output_file = f"{output_prefix}_{model}.pdb"
        output_files.append(output_file)
        
        with open(output_file, 'w') as f:
            # Write header
            f.write(f"REMARK   Generated from {input_pdb}, pose/model {model}\n")
            
            # Write protein atoms
            for line in protein_lines:
                f.write(line)
            
            # Write TER record after protein
            if protein_lines:
                f.write("TER\n")
            
            # Write ligand atoms for this model
            for category in ['ligands', 'cofactors', 'metals', 'waters']:
                for key, atoms in models[model][category].items():
                    for _, line in sorted(atoms, key=lambda x: x[0]):
                        f.write(line)
            
            # Write TER record after ligands and other HETATMs
            f.write("TER\n")
            
            # Write CONECT records for this model
            for line in model_conect_lines[model]:
                f.write(line)
            
            # Write END record
            f.write("END\n")
    
    print(f"Split {len(output_files)} poses from {input_pdb}")
    for file in output_files:
        print(f"  Created: {file}")
    
    return output_files
```
