# Combine Docking Output with original PDB of protein for superimposition

- before we calculate the RMSD we need to combine the diffdock output (SDF) with the Protein (PDB) used for docking
- then we can superimpose proteins
- then we calculate the RMSD between ligands

```python
import pandas as pd
import os
from cg_combine_files import combine_protein_ligand_structures
```

## view overview

```python
# Load your dataframe with protein and ligand information
df = pd.read_csv("/home/ec2-user/SageMaker/data/raw/docking/Experiment_Docking_Overview.csv")
df.head()
```

## combine diffdock result with protein

```python
# Define directories
protein_dir = "/home/ec2-user/SageMaker/data/raw/docking/prep_Flare"
ligand_base_dir = "/home/ec2-user/SageMaker/data/interim/docking/Res_Diffdock_Flare_R1"
output_dir = "/home/ec2-user/SageMaker/data/processed/docking/Res_Diffdock_Flare_R1"


# Run the combination process
results = combine_protein_ligand_structures(
    df=df,
    protein_dir=protein_dir,
    ligand_base_dir=ligand_base_dir,
    output_dir=output_dir,
    protein_col='Protein',  # Customize if your column names are different
    ligand_col='Ligand'
)


# Display summary
print(f"Successfully combined {results['successful']} out of {results['total_combinations']} structures")
```
