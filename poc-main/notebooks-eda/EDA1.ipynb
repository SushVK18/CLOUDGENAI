```python
from Bio import PDB 
import py3Dmol
import ipywidgets as widgets
from IPython.display import display, HTML
```

```python
# Load the PDB file
#pdb_file = 'data/gt/bcsin_261-red.pdb'  # Replace with your PDB file
#pdb_file = 'data/public/1aem.pdb'  # Replace with your PDB file
pdb_file1 = 'data/gt/bcsin_110.pdb'
pdb_file2 = 'data/gt/bcsin_434.pdb'
pdb_file3 = 'data/gt/bcsin_261.pdb'
```

```python
# Visualiye Protein only

# Create a 3Dmol view
view = py3Dmol.view(width=800, height=400)
# Load your PDB file (replace 'your_file.pdb' with your actual PDB file)
view.addModel(open(pdb_file3).read(), 'pdb')
# Set the stick style for all records
view.setStyle({'cartoon': {'color': 'spectrum'}})
# Set the stick style for all HETATM records (ligands)
#view.setStyle({'stick': {'colorscheme': 'set1'}}, 'HETATM')

# Optionally, you can add a background color
view.setBackgroundColor('white')
# Zoom to fit the structure
view.zoomTo()
# Show the visualization
view.show()
```

```python
# Create a 3Dmol view
view = py3Dmol.view(width=1200, height=1200)

# Load your PDB file (replace 'your_file.pdb' with your actual PDB file)
view.addModel(open(pdb_file3).read(), 'pdb')

# Set protein as cartoon representation
#view.setStyle({'protein': {'cartoon': {'color': 'spectrum'}}})
view.setStyle({'cartoon': {'colorscheme': 'whiteCarbon', 'opacity': 0.7}})

# Set ligand as stick representation
# This selects all non-protein, non-water molecules (typical for ligands)
view.setStyle({'hetflag': True, 'not': {'resn': 'HOH'}},
              {'stick': {'colorscheme': 'greenCarbon', 'radius': 0.3}})

# Zoom to fit the structure
view.zoomTo()

# Enable spinning (optional)
# view.spin(True)

# Show the viewer
view.show()
```

```python
# Create a viewer with specific dimensions
view = py3Dmol.view(width=500, height=500)

# Add the PDB data
view.addModel(open(pdb_file3).read(), 'pdb')

# First apply styles to the entire structure
view.setStyle({}, {'cartoon': {'colorscheme': 'whiteCarbon', 'hidden': True}})

# Set protein as cartoon representation - use more specific selection
# 'cartoon' selects backbone atoms, 'protein' selects protein residues
#view.setStyle({'polymer': True, 'resn': {'not': 'HOH'}}, 
#             {'cartoon': {'color': 'spectrum', 'opacity': 0.9}})

# Set ligand as stick representation
# This selects all hetero atoms (non-protein, non-water molecules)
view.setStyle({'hetflag': True, 'not': {'resn': 'HOH'}}, 
             {'stick': {'colorscheme': 'greenCarbon', 'radius': 0.3}})

# Zoom to fit the structure
view.zoomTo()

# Enable spinning (optional)
# view.spin(True)

# Show the viewer
view.show()
```
