```python
import pandas as pd
from uniprot_client import UniProtQuerier
```

```python
# read data from a csv
df = pd.read_csv("raw/docking.csv")
```

```python
df
```

```python
uniprot_ids = df.uniprotID.to_list()
print(f"Total length: {len(uniprot_ids)}")
print(f"Unique Uniprot entries: {len(set(uniprot_ids))}")
```

```python
querier = UniProtQuerier(output_dir="external/uniprot")

# querry by IDs
results = querier.query(
    entry=uniprot_ids  # list of IDs 
)

# Download results
print(f"Total results found: {len(results['results'])}")
downloaded_files = querier.download_results(results, formats=["json", "fasta", "tsv"])

# Print file locations
for fmt, filepath in downloaded_files.items():
    print(f"Downloaded {fmt} file to: {filepath}")
```
